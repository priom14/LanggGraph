```python
from langgraph.graph import StateGraph, START,END
from typing import TypedDict
```

```python
class CricState(TypedDict):
    runs: int
    balls: int
    fours: int
    sixes: int
    
    sr: float
    bpb: float
    boundary_percentage: float
    summary: str
    
    
```

```python
def cal_sr(state: CricState) -> CricState:
    sr = (state['runs']/state['balls'])*100
    
    return {'sr' : sr}
```

```python
def cal_bpb(state: CricState) -> CricState:
    bpb = (state['balls']/(state['fours']+ state['sixes']))
    
    return {'bpb' : bpb}
```

```python
def cal_bp(state: CricState) -> CricState:
    bp = (((state['fours']*4) + (state['sixes'])*6)/state['runs'])*100
    
    return {'boundary_percentage' : bp}
```

```python
def summary(state: CricState) -> CricState:
    state['summary'] = f'''
    The summary of the batter is :
    Runs - {state['runs']}
    Balls faced - {state['balls']}
    
    Strike Rate - {state['sr']}
    Balls Per Boundary - {state['bpb']}
    Boundary Percentage - {state['boundary_percentage']}
    
    '''
    
    return state
```

```python
graph = StateGraph(CricState)

graph.add_node("strike_rate", cal_sr)
graph.add_node("balls_per_boundary", cal_bpb)
graph.add_node("boundary_percentage", cal_bp)
graph.add_node("summary", summary)

graph.add_edge(START, "strike_rate")
graph.add_edge(START, "balls_per_boundary")
graph.add_edge(START, "boundary_percentage")

graph.add_edge("strike_rate", "summary")
graph.add_edge("balls_per_boundary", "summary")
graph.add_edge("boundary_percentage", "summary")

graph.add_edge("summary", END)

workflow = graph.compile()
```

```python
initial_state = {
    'runs' : 100,
    'balls' : 38,
    "fours" : 6,
    "sixes" : 4
}
```

```python
final_state = workflow.invoke(initial_state)
```
